### 2) Create Search Tool for the agent

```python
from langchain.tools import tool

@tool
def search_tool(query: str) -> str:
    """Search information about the query
    Args:
        query: query which is searched
    """
    response = tavily_client.search(query)
    output = ""
    count = 1
    for r in response["results"]:
        output = output +  "{0}- Content: {1}. Citation: {2} \n".format(count, r["content"], r["url"])
        count = count + 1
    return output
```

### 3) Create Deep Agent with Skill

```python
from langchain.chat_models import init_chat_model
from deepagents import create_deep_agent

model = init_chat_model("gpt-4o")

agent = create_deep_agent(
    system_prompt = "Answer directly whenever possible. Avoid asking follow-up questions unless critical information is missing.",
    model=model,
    tools=[search_tool],
    skills=["./skills"]
)
```

### 4) Run Agent

```python
response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "What is ETF?",
            }
        ]
    }
)
print(response["messages"][1].content)
```
