## Authentication Setup

```python
from openai import AsyncOpenAI
from agents import (
    set_default_openai_client,
    set_default_openai_api,
    set_tracing_disabled,
)
from aws_bedrock_token_generator import provide_token

client = AsyncOpenAI(
    api_key=provide_token(),
    base_url="https://bedrock-mantle.us-east-1.api.aws/openai/v1",
    project="default"
)

set_default_openai_client(client)
set_default_openai_api("responses")
set_tracing_disabled(True)  # OpenAI-platform tracing can't reach Mantle
```

## Import Libraries

Import `draw_graph` from the visualization extension:

```python
import asyncio

from agents import Agent, function_tool
from agents.extensions.visualization import draw_graph
```

## Step 1: Define a Function Tool

Create a simple tool that will appear in the visualization:

```python
@function_tool
def get_weather(city: str) -> str:
    return f"The weather in {city} is sunny."
```

## Step 2: Create Multi-Agent System

Build a triage agent with specialist language agents:

💡 **System Design**: Triage pattern routes users to specialists based on language preference.

```python
spanish_agent = Agent(
    name="Spanish agent",
    instructions="You only speak Spanish.",
     model=model_id,
)

english_agent = Agent(
    name="English agent",
    instructions="You only speak English",
     model=model_id,
)
```

## Step 3: Create Orchestrator Agent

Build the triage agent that coordinates handoffs and tools:

```python
triage_agent = Agent(
    name="Triage agent",
    instructions="Handoff to the appropriate agent based on the language of the request.",
    handoffs=[spanish_agent, english_agent],
    tools=[get_weather],
    model=model_id,
)
```

## Step 4: Visualize the Architecture

Use `draw_graph()` to generate an interactive diagram:

**What You'll See**:
- Agents as nodes (Triage agent, Spanish agent, English agent)
- Handoff connections (arrows from Triage to specialists)
- Tools attached to agents (get_weather on Triage agent)

🔍 **Inline Display**: The graph renders directly in the Jupyter notebook!

```python
draw_graph(triage_agent)
```

## Step 5: Export to File

Save the diagram as an image file for documentation:

💡 **Pro Tip**: Use exported diagrams in architecture docs, presentations, or README files.

🎯 **Result**: Creates `agent_graph.png` in the current directory!

```python
draw_graph(triage_agent, filename="agent_graph")
```
