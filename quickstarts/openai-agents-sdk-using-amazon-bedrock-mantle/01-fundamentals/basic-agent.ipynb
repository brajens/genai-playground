## Step 2: Configure Authentication

Set up the OpenAI client to connect to AWS Bedrock Mantle endpoints.

**Key points:**
- `provide_token()` generates AWS credentials automatically
- Base URL points to Bedrock Mantle in us-east-1
- Tracing is disabled because OpenAI platform can't reach Mantle

```python
from openai import AsyncOpenAI
from agents import (
    set_default_openai_client,
    set_default_openai_api,
    set_tracing_disabled,
)
from aws_bedrock_token_generator import provide_token

client = AsyncOpenAI(
    api_key=provide_token(),
    base_url="https://bedrock-mantle.us-east-1.api.aws/openai/v1",
    project="default"
)

set_default_openai_client(client)
set_default_openai_api("responses")
set_tracing_disabled(True)  # OpenAI-platform tracing can't reach Mantle
```

## Step 3: Import Agent Framework

Import the core classes needed to create and run agents.

```python
import asyncio
from agents import Agent, Runner
```

## Step 4: Create Your First Agent

Define an agent with:
- **name**: Identifies the agent (useful in multi-agent systems)
- **instructions**: Tells the agent how to behave and respond
- **model**: The Mantle model ID to use ("openai.gpt-5.5")

💡 **Tip**: Clear, specific instructions lead to better agent behavior!

```python
agent = Agent(
    name="History tutor",
    instructions="You answer history questions clearly and concisely.",
    model="openai.gpt-5.5",  # Mantle model ID
)
```

## Step 5: Run the Agent

Use `Runner.run()` to execute the agent with user input.

**How it works:**
1. `Runner.run()` sends your input to the agent
2. The agent processes it using the model and instructions
3. Returns a result object with `final_output` containing the response

⚡ **Note**: This is an async operation, use `await` in Jupyter notebooks.

```python
result = await Runner.run(agent, "When did the Roman Empire fall?")
print(result.final_output)
```
