## Import Libraries

Import `ModelSettings` and `Reasoning` for configuration:

```python
import asyncio

from agents import Agent, Runner, ModelSettings
from openai.types.shared import Reasoning
```

## Step 1: Configure Model Settings

Use `ModelSettings` to control reasoning effort and verbosity:

**Reasoning Effort Levels**:
- `"low"`: Faster, less compute, good for simple tasks
- `"medium"`: Balanced (default)
- `"high"`: More thorough, slower, better for complex tasks

**Verbosity Levels**:
- `"low"`: Concise responses
- `"medium"`: Balanced detail (default)
- `"high"`: Detailed, explanatory responses

💡 **Pro Tip**: Use `low` for chat/quick queries, `high` for analysis/reasoning tasks.

```python
agent = Agent(
    name="My Agent",
    instructions="You're a helpful agent.",
    model=model_id,
    model_settings=ModelSettings(
        reasoning=Reasoning(effort="low"),  # Fast, less compute
        verbosity="low"  # Concise responses
    )
)
```

## Step 2: Run and Compare

Test the agent with low settings:

🎯 **Result**: Fast, concise response!

```python
result = await Runner.run(agent, "How can you help me?")
print(result.final_output)
```
