```python
# Import libraries
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
# Model used for testing
model_name = "google/flan-t5-small"
```

```python
# Loads the correct tokenizer for the chosen Hugging Face mode
tokenizer = AutoTokenizer.from_pretrained(model_name)

# loads the model in full precision
model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    device_map="auto" 
)
```

```python
# Count the parameters of the model
total_params = sum(p.numel() for p in model.parameters())
print(f"Total params: {total_params:,}")
```

```python
# parameter data type = FP32 which takes 32 bits or 4 bytes
# Measure size of the model
model_size = (total_params * 4)/(1024*1024)
print(f"Model size is: {model_size:,} in MB")
```

```python
# Function to used to call the model
def call_model(input_text,the_model,the_tokenizer):
    inputs = the_tokenizer(input_text, return_tensors="pt")
    outputs = the_model.generate(**inputs)
    result = the_tokenizer.decode(outputs[0], skip_special_tokens=True)
    return result
```

```python
prompt = "translate English to French: How are you?"
```

```python
# Call model with full precision
print(call_model(prompt,model,tokenizer))
```

```python
# Load model in float16 (half precision)
small_model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    dtype=torch.float16,
    device_map="auto" 
)
```

```python
# Count the parameters again. You will find no changes.
total_params = sum(p.numel() for p in small_model.parameters())
print(f"Total params: {total_params:,}")
```

```python
# Call model with small precision
print(call_model(prompt,small_model,tokenizer))
```
