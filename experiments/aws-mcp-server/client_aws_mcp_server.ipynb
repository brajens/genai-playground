```python
import asyncio
import uuid

from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client
```

```python
# MCP Endpoint - using end point of eu-east-1 region
MCP_ENDPOINT = "https://aws-mcp.us-east-1.api.aws/mcp"
AWS_REGION = "us-east-1"
```

### 1) List AWS MCP Server Tools

```python
async def main():

    async with streamablehttp_client(
        MCP_ENDPOINT
    ) as (read, write, _):

        async with ClientSession(read, write) as session:

            # Initialize MCP session
            await session.initialize()

            # List tools exposed by AWS MCP
            tools = await session.list_tools()

            print("AVAILABLE TOOLS:\n")

            for tool in tools.tools[:20]:
                print("-", tool.name)
```

```python
await main()
```

### 2) Search into AWS Documentation

```python
QUERY = """
What is the hourly cost of a t3.medium EC2 instance
in us-east-1 Linux on-demand pricing?
"""

async def main():

    async with streamablehttp_client(
        MCP_ENDPOINT
    ) as (read, write, _):

        async with ClientSession(read, write) as session:

            # Initialize MCP session
            await session.initialize()

            result = await session.call_tool(
                "aws___search_documentation",
                {
                    "search_phrase": QUERY
                }
            )

            print("\nDOCUMENTATION RESULT:\n")

            # MCP responses are often structured content
            for item in result.content:
                print(item.text)
```

```python
await main()
```

## 3) Check Regional Availability of AWS Service

```python
async def main():

    async with streamablehttp_client(
        MCP_ENDPOINT
    ) as (read, write, _):

        async with ClientSession(read, write) as session:

            await session.initialize()

            result = await session.call_tool(
                "aws___get_regional_availability",
                {
                    "resource_type": "product",
                    "filters": ["Amazon Bedrock"],
                    "regions": ["ap-south-1"]
                }
            )

            print(result)
```

```python
await main()
```
